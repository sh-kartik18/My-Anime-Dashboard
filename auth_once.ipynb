```python
import requests

CLIENT_ID = "" #Enter your Anilist API Client ID
CLIENT_SECRET = "" #Enter your Anilist API Client Secret
REDIRECT_URI = "https://anilist.co/api/v2/oauth/pin"

# Step 1: Ask user to visit this URL and get the code
auth_url = f"https://anilist.co/api/v2/oauth/authorize?client_id={CLIENT_ID}&redirect_uri={REDIRECT_URI}&response_type=code"
print("Go to this URL and log in:", auth_url)

# Step 2: Paste the code here
code = input("Enter the code from AniList: ")

# Step 3: Exchange code for access + refresh tokens
token_url = "https://anilist.co/api/v2/oauth/token"
payload = {
    "grant_type": "authorization_code",
    "client_id": CLIENT_ID,
    "client_secret": CLIENT_SECRET,
    "redirect_uri": REDIRECT_URI,
    "code": code
}
response = requests.post(token_url, json=payload).json()
print(response)

access_token = response["access_token"]
refresh_token = response["refresh_token"]

# Save tokens for future
with open("tokens.json", "w") as f:
    import json
    json.dump(response, f, indent=2)

print("Tokens saved to tokens.json")
```
